### Comparação de duas variâncias: Teste F Bilateral

O teste F compara as variâncias de duas populações. Ele costuma ser feito antes do teste t para duas médias independentes, para decidir se as variâncias podem ser tratadas como iguais (homocedasticidade) ou diferentes (heterocedasticidade). Partimos de duas amostras aleatórias e independentes, de tamanhos n1 e n2, de populações normais com variâncias σ1² e σ2².


#### Desejamos testar a hipótese bilateral:

    H0: σ1² = σ2²  Ha: σ1² ≠ σ2²

#### Condições
As duas populações devem ter distribuição normal.

#### Estatística de teste

$$F = \frac{s_1^2}{s_2^2}$$

#### Notação
    n1 = tamanho da 1ª amostra
    n2 = tamanho da 2ª amostra
    σ1² = variância populacional da 1ª amostra
    σ2² = variância populacional da 2ª amostra
    s1²= variância amostral da 1ª amostra
    s2²= variância amostral da 2ª amostra

A estatística F, razão entre as duas variâncias amostrais, segue a distribuição F de Fisher-Snedecor, com n1 – 1 graus de liberdade no numerador e n2 – 1 no denominador.

Para um teste bilateral, a regra de decisão para um nível de significância (α) é:

    Rejeitar H0 se: F < FL ou se F > FR

onde os valores críticos são obtidos por:

$$F_L = \frac{1}{F_{n_2 - 1; n_1 - 1; \alpha/2}} \quad \text{e} \quad F_R = F_{n_1 - 1; n_2 - 1; \alpha/2}$$

em que $F_{a; b; \alpha/2}$ é o valor que deixa área $\alpha/2$ na cauda superior da distribuição F com $a$ graus de liberdade no numerador e $b$ no denominador.

Para um teste bilateral, a regra de decisão para um nível de significância (α) é:

    Rejeitar H0 se: F < FL ou se F > FR

onde os valores críticos são obtidos por:

$$F_L = \frac{1}{F_{n_2 - 1; n_1 - 1; \alpha/2}} \quad \text{e} \quad F_R = F_{n_1 - 1; n_2 - 1; \alpha/2}$$

em que $F_{a; b; \alpha/2}$ é o valor que deixa área $\alpha/2$ na cauda superior da distribuição F com $a$ graus de liberdade no numerador e $b$ no denominador.

#### Exemplo
Uma fábrica de chocolates comprou uma máquina nova de embalagem e quer saber se a variabilidade do peso das barras mudou em relação à máquina antiga. Foi tomada uma amostra aleatória de cada máquina (dados fictícios):

    Máquina antiga: n1 = 13 ; ̅𝑥̅1= 100,4 g ; s1² = 0,8 g²
    Máquina nova: n2 = 11 ; ̅𝑥̅2= 100,1 g ; s2² = 3,2 g²

Vamos fazer o teste bilateral de comparação das variâncias com nível de significância de 5%.

Hipóteses estatísticas: H0: σ1² = σ2²  Ha: σ1² ≠ σ2²

Estatística de teste:

$$F = \frac{s_1^2}{s_2^2} = \frac{0{,}8}{3{,}2} = 0{,}25$$

Para obter os valores críticos devemos consultar a tabela F:

Com $n_1 - 1 = 12$ e $n_2 - 1 = 10$ graus de liberdade e área $\alpha/2 = 0{,}025$ na cauda superior (valores conferidos com o `scipy` na célula abaixo):

$$F_{10; 12; 0{,}025} \approx 3{,}374 \qquad F_{12; 10; 0{,}025} \approx 3{,}621$$

```python
# Valores críticos da distribuição F (área 0,025 na cauda superior)
from scipy.stats import f

print("F(10; 12) =", round(f.ppf(1 - 0.025, 10, 12), 3))
print("F(12; 10) =", round(f.ppf(1 - 0.025, 12, 10), 3))
```

Assim, encontramos:

$$F_L = \frac{1}{F_{10; 12; 0{,}025}} = \frac{1}{3{,}374} \approx 0{,}296 \qquad F_R = F_{12; 10; 0{,}025} \approx 3{,}621$$

Com α = 0,05, a regra de decisão fica:
Rejeitar H0 se: F < 0,296 ou se F > 3,621

Como F = 0,25 < 0,296, rejeitamos H0: ao nível de 5% de significância, a variância do peso das barras é diferente nas duas máquinas.

```python
#Realizando o teste de hipósteses bilateral para comparação de duas variâncias.
import numpy as np
import scipy.stats

#dados = [1,3,6,0]


#Definindo variáveis do enunciado:

N_amostras_A = 13 #número de amostras
Variância_A = 0.8 #variância = desvio_padrão²

#Desvio_Padrão = X
#Variância_dados_A = np.var(dados)

N_amostras_B = 11 #número de amostras
Variância_B = 3.2 #variância = desvio_padrão²

Significância = 5 #% (alfa)



#Calculando variáveis secundárias:
Alfa = Significância/100
Meio_Alfa = Alfa/2
Meio_Grau = 1-Meio_Alfa

Valor_crítico_L = scipy.stats.f.ppf(q=Meio_Grau, dfn=N_amostras_B-1, dfd=N_amostras_A-1)
Valor_crítico_R = scipy.stats.f.ppf(q=Meio_Grau, dfn=N_amostras_A-1, dfd=N_amostras_B-1)

FL = round(1/Valor_crítico_L, 4)
FR = round(Valor_crítico_R, 4)

Estat_teste = round(Variância_A/Variância_B, 4)

print("Alfa: ", Alfa)
print("Estatística de teste: ", Estat_teste)
print("Valor_crítico_L: ", Valor_crítico_L)
print("Valor_crítico_R: ", Valor_crítico_R)
print("\n")


#Verificando a hipótese:
if Estat_teste < FL or Estat_teste > FR:
    print("H0 rejeitada (Heterocedásticas), pois {0} fora do intervalo de FL {1} a FR {2}" .format(Estat_teste, FL, FR))
else:
    print("H0 aceita (Homocedásticas), pois {0} dentro do intervalo de FL {1} a FR {2}" .format(Estat_teste, FL, FR))
```

### Comparação de duas médias independentes: Teste T-Student

Uma pergunta comum é: "o método A dá resultados diferentes do método B?". Em termos estatísticos, comparamos as médias de dois grupos, medidos após a aplicação de cada método. Quando as variâncias populacionais são desconhecidas, a ferramenta mais usada para isso é o teste t de Student para duas amostras.

Vamos comparar as médias de duas populações independentes, ambas normais, com uma amostra aleatória de tamanho n1 da população 1 e outra de tamanho n2 da população 2.

#### Suposições
1. As duas amostras são independentes;
2. As duas amostras vêm de populações com distribuição normal. Se a normalidade não se confirmar, uma alternativa são os testes não paramétricos, que não dependem da distribuição dos dados.

#### Condições
Populações normais, com variâncias populacionais desconhecidas.

Há dois casos: variâncias desconhecidas mas iguais (homocedasticidade) e variâncias desconhecidas e diferentes (heterocedasticidade).

#### Caso 1: As duas populações parecem ter variâncias iguais (Homocedásticas):
#### Notação utilizada:

    𝑥̅1 = média da 1ª amostra
    𝑥̅2 = média da 2ª amostra
    n1 = tamanho da 1ª amostra
    n2 = tamanho da 2ª amostra
    s1² = variância da 1ª amostra
    s2² = variância da 2ª amostra

#### Hipóteses:

    H0: μ1 = μ2 versus Ha: μ1 ≠ μ2 → Hipótese bilateral
    H0: μ1 <= μ2 versus Ha: μ1 > μ2 → Hipótese unilateral à direita
    H0: μ1 >= μ2 versus Ha: μ1 < μ2 → Hipótese unilateral à esquerda

#### Estatística de teste:

$$t_0 = \frac{\bar x_1 - \bar x_2}{\sqrt{\dfrac{s_{\text{comb}}^2}{n_1} + \dfrac{s_{\text{comb}}^2}{n_2}}}$$

Como σ² é desconhecida, ela precisa ser estimada. As duas variâncias amostrais, s1² e s2², estimam σ² sem viés; combinamos as duas numa só estimativa, a variância combinada:

$$s_{\text{comb}}^2 = \frac{(n_1 - 1) s_1^2 + (n_2 - 1) s_2^2}{(n_1 - 1) + (n_2 - 1)}$$

#### Valor-p:

| Tipo de teste | Como obter o valor p |
|---|---|
| Unilateral à direita | área à direita da estatística de teste |
| Bilateral | duas vezes a área à direita do valor absoluto da estatística de teste |
| Unilateral à esquerda | área à esquerda da estatística de teste |

Concluindo um teste de hipótese utilizando o valor p:
-  Se valor p > α, então aceitamos H0;
-  Se valor p <= α, então rejeitamos H0.

#### Número de graus de liberdade:
Quando as variâncias amostrais são combinadas: gl = n1 + n2 – 2

#### Intervalo de confiança:

$$(\bar x_1 - \bar x_2) \pm t_{gl; \alpha/2} \cdot \sqrt{\frac{s_{\text{comb}}^2}{n_1} + \frac{s_{\text{comb}}^2}{n_2}}$$

Como usar o intervalo para concluir:
- As médias são iguais ou diferentes? O teste bilateral equivale a

    H0: μ1 - μ2 = 0 versus Ha: μ1 - μ2 ≠ 0
    
    Se o intervalo contém 0 (zero), os dados são compatíveis com médias iguais; se não contém, conclui-se, com 100(1-α)% de confiança, que as médias são diferentes.
    
    
- Se forem diferentes, qual é maior?
    
    Se os dois limites do intervalo são negativos (-), a média da 2ª amostra é a maior.
   
    Se os dois limites são positivos (+), a média da 1ª amostra é a maior.

#### Caso 2: As duas populações parecem ter variâncias desiguais (Heterocedásticas):
#### Notação utilizada:

    𝑥̅1 = média da 1ª amostra
    𝑥̅2 = média da 2ª amostra
    n1 = tamanho da 1ª amostra
    n2 = tamanho da 2ª amostra
    s1² = variância da 1ª amostra
    s2² = variância da 2ª amostra

#### Hipóteses:

    H0: μ1 = μ2 versus Ha: μ1 ≠ μ2 → Hipótese bilateral
    H0: μ1 <= μ2 versus Ha: μ1 > μ2 → Hipótese unilateral à direita
    H0: μ1 >= μ2 versus Ha: μ1 < μ2 → Hipótese unilateral à esquerda

#### Estatística de teste:

$$t_0 = \frac{\bar x_1 - \bar x_2}{\sqrt{\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}}}$$

#### Valor-p:

| Tipo de teste | Como obter o valor p |
|---|---|
| Unilateral à direita | área à direita da estatística de teste |
| Bilateral | duas vezes a área à direita do valor absoluto da estatística de teste |
| Unilateral à esquerda | área à esquerda da estatística de teste |

Concluindo um teste de hipótese utilizando o valor p:
-  Se valor p > α, então aceitamos H0;
-  Se valor p <= α, então rejeitamos H0.

#### Número de graus de liberdade:
Número de graus de liberdade: o menor dos dois n1 – 1 e n2 – 1.

#### Intervalo de confiança (somente para hipóteses bilaterais):

$$(\bar x_1 - \bar x_2) \pm t_{gl; \alpha/2} \cdot \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}$$

A conclusão do teste de hipóteses utilizando o valor p ou o intervalo de confiança deve ser feita da mesma maneira como descrito no Caso 1.

#### Exemplo 1:
Uma escola oferece um curso de Excel em duas modalidades: presencial (P) e online (O). Para comparar o desempenho, alunos de cada turma foram sorteados e cronometrados resolvendo a mesma tarefa. Os tempos, em minutos, estão resumidos abaixo (dados fictícios):

| Turma | $n$ | Média | Desvio padrão |
|---|---|---|---|
| P | 15 | 18,4 | 5,2 |
| O | 12 | 23,9 | 5,6 |

Ao nível de 5% de significância, há evidência de que o tempo médio das duas turmas é diferente?

Como vamos comparar duas médias, primeiro verificamos se as variâncias podem ser consideradas iguais, com o teste F bilateral:

#### Testando F bilateral:

Hipóteses estatísticas: H0: σ1² = σ2² versus Ha: σ1² ≠ σ2² → Hipótese bilateral

Estatística de teste:

$$F = \frac{s_P^2}{s_O^2} = \frac{5{,}2^2}{5{,}6^2} \approx 0{,}8622$$

Para obter os valores críticos devemos consultar a tabela F. Com $n_P = 15$ e $n_O = 12$ e área $\alpha/2 = 0{,}025$ na cauda superior (valores conferidos com o `scipy` na célula abaixo):

$$F_L = \frac{1}{F_{11; 14; 0{,}025}} = \frac{1}{3{,}095} \approx 0{,}3231 \qquad F_R = F_{14; 11; 0{,}025} \approx 3{,}359$$

```python
# Valores críticos da distribuição F (área 0,025 na cauda superior)
from scipy.stats import f

print("F(11; 14) =", round(f.ppf(1 - 0.025, 11, 14), 3))
print("F(14; 11) =", round(f.ppf(1 - 0.025, 14, 11), 3))
```

Com α = 0,05, a regra de decisão é:

Rejeitar H0 se: F < 0,3231 ou se F > 3,359

Como F = 0,8622, NÃO rejeitamos H0: ao nível de 5%, não há evidência de diferença entre as variâncias dos tempos das duas turmas (tratamos as variâncias como iguais).

Podemos então fazer o teste t para duas médias, supondo variâncias iguais.

#### Testando T para duas médias independentes homocedásticas

Hipóteses:  H0: μ1 = μ2 versus Ha: μ1 ≠ μ2 → Hipótese bilateral

Estatística de teste:

$$t_0 = \frac{\bar x_P - \bar x_O}{\sqrt{\dfrac{s_{\text{comb}}^2}{n_P} + \dfrac{s_{\text{comb}}^2}{n_O}}} = \frac{18{,}4 - 23{,}9}{\sqrt{\dfrac{28{,}9408}{15} + \dfrac{28{,}9408}{12}}} \approx -2{,}6397$$

A variância comum σ² foi estimada pela variância combinada:

$$s_{\text{comb}}^2 = \frac{(15 - 1) \cdot 5{,}2^2 + (12 - 1) \cdot 5{,}6^2}{(15 - 1) + (12 - 1)} = 28{,}9408$$

Valor-p:

Como o teste é bilateral, o valor p é 2 x a área à direita do módulo da estatística de teste.

Com variâncias combinadas, os graus de liberdade são gl = n1 + n2 – 2 = 15 + 12 – 2 = 25.

Linha de 25 graus de liberdade da tabela t-Student (calculada com o `scipy` na célula abaixo):

| Área na cauda superior | 0,25 | 0,125 | 0,10 | 0,05 | 0,025 | 0,0125 | 0,01 | 0,005 |
|---|---|---|---|---|---|---|---|---|
| $t$ (25 g.l.) | 0,684 | 1,178 | 1,316 | 1,708 | 2,060 | 2,385 | **2,485** | **2,787** |

O valor $\lvert t_0 \rvert = 2{,}6397$ fica entre 2,485 (coluna 0,01) e 2,787 (coluna 0,005); logo, pela tabela, $P(T > 2{,}6397)$ está entre 0,005 e 0,01, e o valor p, entre 0,01 e 0,02. O cálculo exato na célula abaixo dá $P(T > 2{,}6397) \approx 0{,}0070$, ou seja, valor p $\approx 0{,}0141$. Como o valor p < α, rejeitamos a hipótese nula.

Concluímos que o tempo médio para concluir a tarefa é diferente nas duas turmas.

```python
# Linha de 25 graus de liberdade da tabela t-Student e valor p exato do teste
from scipy.stats import t

areas = [0.25, 0.125, 0.10, 0.05, 0.025, 0.0125, 0.01, 0.005]
for area in areas:
    print(f"área {area}: t = {t.ppf(1 - area, 25):.3f}")

t0 = -2.6397
print(f"\nP(T > |t0|) = {t.sf(abs(t0), 25):.4f}")
print(f"valor p (bilateral) = {2 * t.sf(abs(t0), 25):.4f}")
```

```python
#Realizando teste T para duas médias independentes:
import numpy as np
import scipy.stats


#dados_A = [1,3,6,0]
#Média_dados_A = np.mean(dados_A)
#Variância_dados_A = np.var(dados_A)


#Definindo variáveis do enunciado:

# Amostra 1 (A no código): turma presencial; amostra 2 (B no código): turma online, como no texto (F = s_P² / s_O²)
N_amostras_A = 15 #número de amostras
Média_amostral_A = 18.4 #média das amostras
Desvio_Padrão_A = 5.2   #Variância_A**0.5
Variância_A = Desvio_Padrão_A**2

N_amostras_B = 12 #número de amostras
Média_amostral_B = 23.9 #média das amostras
Desvio_Padrão_B = 5.6   #Variância_B**0.5
Variância_B = Desvio_Padrão_B**2

Significância = 5 #% (alfa)

# Descrição das hipóteses:

H0 = "Média_A_=_Média_B"   #Hipótese nula (com igualdade)
Ha = "Média_A_<>_Média_B"   #Hipótese alternativa (sem igualdade)




#Realizando o teste de hipósteses bilateral para comparação de duas variâncias.

#Calculando variáveis secundárias:
Alfa = Significância/100
Meio_Alfa = Alfa/2
Meio_Grau = 1-Meio_Alfa

Valor_crítico_L = scipy.stats.f.ppf(q=Meio_Grau, dfn=N_amostras_B-1, dfd=N_amostras_A-1)
Valor_crítico_R = scipy.stats.f.ppf(q=Meio_Grau, dfn=N_amostras_A-1, dfd=N_amostras_B-1)

FL = round(1/Valor_crítico_L, 4)
FR = round(Valor_crítico_R, 4)

Estat_teste = round(Variância_A/Variância_B, 4)

print("------------------------------TESTE F----------------------------------")
print("Alfa: ", Alfa)
print("Estatística de teste: ", Estat_teste)
print("Valor_crítico_L: ", Valor_crítico_L)
print("Valor_crítico_R: ", Valor_crítico_R)
print("\n")


#Verificando a hipótese:
if Estat_teste < FL or Estat_teste > FR:
    print("H0 rejeitada (Heterocedásticas), pois {0} fora do intervalo de FL {1} a FR {2}" .format(Estat_teste, FL, FR))
    print("----------------------------------------------------------------------")
    Homocedásticas = False
else:
    print("H0 aceita (Homocedásticas), pois {0} dentro do intervalo de FL {1} a FR {2}" .format(Estat_teste, FL, FR))
    print("----------------------------------------------------------------------")
    Homocedásticas = True



#Realizando o teste de hipósteses para as Médias.

print("------------------------------TESTE T----------------------------------")
print("H0: ",H0)
print("Ha: ",Ha)
print("\n")

#Cálculo da estatística de teste e valor-p:
Estat_teste, P_valor = scipy.stats.ttest_ind_from_stats(mean1=Média_amostral_A, std1=Desvio_Padrão_A, nobs1=N_amostras_A,
                                            mean2=Média_amostral_B, std2=Desvio_Padrão_B, nobs2=N_amostras_B, equal_var=Homocedásticas)

Estat_teste = round(Estat_teste,4)
P_valor = round(P_valor,4)

print("Estatística de teste: ",Estat_teste)
print("Valor P: ",P_valor)


#Verificação da hipótese nula de acordo com o valor-p:       
print("\n")
print("Conclusão de acordo com a relação do valor-p e a significância:")

if P_valor < Alfa:
    print("H0 rejeitada, pois P-valor {0} < Alfa {1}" .format(P_valor, Alfa))
else:
    print("H0 aceitada, pois P-valor {0} >= Alfa {1}" .format(P_valor, Alfa))
```

#### Exemplo 2:
Uma imobiliária quer comparar o preço médio do metro quadrado (em R\$ 1.000,00) de apartamentos em dois bairros. Foram sorteados anúncios em cada bairro, com os resumos abaixo (dados fictícios):

| Bairro | $n$ | Média | Desvio padrão |

| Norte | 18 | 9,4 | 0,8 |
| Sul | 14 | 7,9 | 2,1 |

```python
#Realizando teste T Bilateral para duas médias independentes:
import numpy as np
import scipy.stats


#dados_A = [1,3,6,0]
#Média_dados_A = np.mean(dados_A)
#Variância_dados_A = np.var(dados_A)


#Definindo variáveis do enunciado:

N_amostras_A = 18 #número de amostras
Média_amostral_A = 9.4 #média das amostras
Desvio_Padrão_A = 0.8   #Variância_A**0.5
Variância_A = Desvio_Padrão_A**2

N_amostras_B = 14 #número de amostras
Média_amostral_B = 7.9 #média das amostras
Desvio_Padrão_B = 2.1   #Variância_B**0.5
Variância_B = Desvio_Padrão_B**2

Significância = 5 #% (alfa)

# Descrição das hipóteses:

H0 = "Média_A_=_Média_B"   #Hipótese nula (com igualdade)
Ha = "Média_A_<>_Média_B"   #Hipótese alternativa (sem igualdade)




#Realizando o teste de hipósteses bilateral para comparação de duas variâncias.

#Calculando variáveis secundárias:
Alfa = Significância/100
Meio_Alfa = Alfa/2
Meio_Grau = 1-Meio_Alfa

Valor_crítico_L = scipy.stats.f.ppf(q=Meio_Grau, dfn=N_amostras_B-1, dfd=N_amostras_A-1)
Valor_crítico_R = scipy.stats.f.ppf(q=Meio_Grau, dfn=N_amostras_A-1, dfd=N_amostras_B-1)

FL = round(1/Valor_crítico_L, 4)
FR = round(Valor_crítico_R, 4)

Estat_teste = round(Variância_A/Variância_B, 4)

print("------------------------------TESTE F----------------------------------")
print("Alfa: ", Alfa)
print("Estatística de teste: ", Estat_teste)
print("Valor_crítico_L: ", Valor_crítico_L)
print("Valor_crítico_R: ", Valor_crítico_R)
print("\n")


#Verificando a hipótese:
if Estat_teste < FL or Estat_teste > FR:
    print("H0 rejeitada (Heterocedásticas), pois {0} fora do intervalo de FL {1} a FR {2}" .format(Estat_teste, FL, FR))
    print("----------------------------------------------------------------------")
    Homocedásticas = False
else:
    print("H0 aceita (Homocedásticas), pois {0} dentro do intervalo de FL {1} a FR {2}" .format(Estat_teste, FL, FR))
    print("----------------------------------------------------------------------")
    Homocedásticas = True



#Realizando o teste de hipósteses para as Médias.

print("------------------------------TESTE T----------------------------------")
print("H0: ",H0)
print("Ha: ",Ha)
print("\n")

#Cálculo da estatística de teste e valor-p:
Estat_teste, P_valor = scipy.stats.ttest_ind_from_stats(mean1=Média_amostral_A, std1=Desvio_Padrão_A, nobs1=N_amostras_A,
                                            mean2=Média_amostral_B, std2=Desvio_Padrão_B, nobs2=N_amostras_B, equal_var=Homocedásticas)

Estat_teste = round(Estat_teste,4)
P_valor = round(P_valor,4)

print("Estatística de teste: ",Estat_teste)
print("Valor P: ",P_valor)


#Verificação da hipótese nula de acordo com o valor-p:       
print("\n")
print("Conclusão de acordo com a relação do valor-p e a significância:")

if P_valor < Alfa:
    print("H0 rejeitada, pois P-valor {0} < Alfa {1}" .format(P_valor, Alfa))
else:
    print("H0 aceitada, pois P-valor {0} >= Alfa {1}" .format(P_valor, Alfa))
```

### Comparação de duas médias pareadas: Teste T-Student

```python
#Realizando teste T Bilateral para duas médias independentes:
import numpy as np
import scipy.stats


#Definindo variáveis do enunciado:

dados_A = [200,174,198,170,179]
dados_B = [191,170,177,167,159]

Significância = 5 #% (alfa)
Alfa = Significância/100

# Descrição das hipóteses:

H0 = "Média_A_=_Média_B"   #Hipótese nula (com igualdade)
Ha = "Média_A_<>_Média_B"   #Hipótese alternativa (sem igualdade)



#Realizando o teste de hipósteses para as Médias.

print("------------------------------TESTE T----------------------------------")
print("H0: ",H0)
print("Ha: ",Ha)
print("\n")

#Cálculo da estatística de teste e valor-p:
Estat_teste, P_valor = scipy.stats.ttest_rel(dados_A,dados_B)

Estat_teste = round(Estat_teste,4)
P_valor = round(P_valor,4)

print("Estatística de teste: ",Estat_teste)
print("Valor P: ",P_valor)


#Verificação da hipótese nula de acordo com o valor-p:       
print("\n")
print("Conclusão de acordo com a relação do valor-p e a significância:")

if P_valor < Alfa:
    print("H0 rejeitada, pois P-valor {0} < Alfa {1}" .format(P_valor, Alfa))
else:
    print("H0 aceitada, pois P-valor {0} >= Alfa {1}" .format(P_valor, Alfa))
```
