# Estatística Geral

```python
DATASET = [15, 5, 3, 8, 10, 2, 7, 11, 12] # Dados para uso em exercícios
```

## Distribuição de frequência


### Rol:

Organizar os dados em ordem crescente ou decrescente.

```python
DATASET.sort() # Ordena os dados em ordem crescente
DATASET
```

```python
DATASET.sort(reverse=True) # Ordena os dados em ordem decrescente
DATASET
```

### Amplitude total:

Diferença entre o maior e o menor valor observado. É a forma mais simples de medir o quanto os dados se espalham:

```python
AT = max(DATASET) - min(DATASET) # Amplitude total = máximo - mínimo
AT
```

### Número de classes de uma distribuição de frequência (K):

Numa tabela de frequências agrupada, os dados são repartidos em faixas chamadas classes. Existem várias regras para escolher quantas classes usar; aqui adotamos a mais simples, K = √n, em que n é o número de observações.

```python
K = len(DATASET) ** (1/2) # Número de classes
K
```

### Amplitude do intervalo de classe (Ac):

É a largura de cada classe: a amplitude total dividida pelo número de classes, Ac = AT / K.

```python
AC = (max(DATASET) - min(DATASET)) / (len(DATASET) ** (1/2)) # Amplitude do intervalo de classe
AC
```

## Medidas de tendência central


```python
import numpy as np
from scipy import stats
```

```python
ARRAY = np.array(DATASET)
```

### Média aritmética:

É a soma de todos os valores observados dividida pela quantidade de valores. Para n observações $x_1, x_2, \ldots, x_n$:

$$\bar{x} = \frac{1}{n} \sum_{i=1}^{n} x_i$$

```python
MEDIA = np.mean(ARRAY)
MEDIA
```

### Mediana:

Com os dados em ordem crescente, a mediana é o valor do meio: metade das observações fica abaixo dela e metade acima. Quando n é par, usa-se a média dos dois valores centrais.

```python
MEDIANA = np.median(ARRAY)
MEDIANA
```

### Moda:

É o valor que aparece mais vezes no conjunto. Ela é útil, por exemplo, quando os dados se concentram nos extremos e a média e a mediana acabam caindo numa região com poucas observações. Um conjunto pode ter mais de uma moda ou nenhuma: no `DATASET` acima todos os valores aparecem uma única vez, e nesse caso o `scipy` devolve o menor deles.

```python
MODA = stats.mode(ARRAY)
MODA
```

## Medidas de dispersão


```python
import numpy as np
from scipy import stats
```

```python
ARRAY = np.array(DATASET)
```

### Variância:

Mede o quanto os valores se afastam da média: é a média dos quadrados dos desvios em relação à média. Para uma população inteira, a soma dos quadrados dos desvios é dividida por n; para uma amostra, divide-se por n − 1 (os graus de liberdade), porque a própria média foi estimada com os mesmos dados. A função `np.var` divide por n por padrão; para a versão amostral, use `np.var(ARRAY, ddof=1)`.

```python
VARIANCIA = np.var(ARRAY)
VARIANCIA
```

### Desvio padrão:

A variância fica na unidade dos dados ao quadrado (metros², reais², ...), o que atrapalha a interpretação. Tirando a raiz quadrada da variância voltamos à unidade original: esse é o desvio padrão. Assim como `np.var`, `np.std` divide por n; use `ddof=1` para o desvio padrão amostral.

```python
DESVIO = np.std(ARRAY)
DESVIO
```

### Coeficiente de variação:

Para comparar a dispersão de conjuntos com médias muito diferentes, ou medidos em unidades diferentes, usa-se uma medida relativa: o coeficiente de variação, CV = (desvio padrão / média) × 100%. Ele não tem unidade e diz quanto o desvio padrão representa em relação à média.

```python
CV = (np.std(ARRAY) / np.mean(ARRAY)) * 100
CV
```

### Escore z (escore padronizado):

O escore z indica a quantos desvios padrão um valor está acima (z > 0) ou abaixo (z < 0) da média: z = (x − média) / desvio padrão. Como regra prática, valores com z entre −2 e 2 são considerados comuns, e valores fora dessa faixa, incomuns.

```python
SZ = (5 - np.mean(ARRAY)) / np.std(ARRAY)
SZ
```

## Medidas de tendência não central


```python
import numpy as np
from scipy import stats
```

```python
ARRAY = np.array(DATASET)
```

### Quantis:

O quantil de ordem p (0 < p < 1), Qp, é o valor que deixa aproximadamente 100p% das observações abaixo dele e 100(1 − p)% acima. Quartis, decis e percentis são casos particulares (p = 0,25, 0,1, 0,01, ...).

```python
QUANTIL = np.percentile(ARRAY, 50) # exemplo de cálculo do segundo quartil que é igual a mediana
QUANTIL
```

### Quartis:

São os três valores que dividem os dados ordenados em quatro partes com o mesmo número de observações: Q1 (25%), Q2 (50%, a mediana) e Q3 (75%).

```python
Q1 = np.percentile(ARRAY, 25) # exemplo de cálculo do primeiro quartil
Q1
```

```python
Q3 = np.percentile(ARRAY, 75) # exemplo de cálculo do terceiro quartil
Q3
```

### Intervalo interquartil:

É a diferença Q3 − Q1. Corresponde à faixa que contém os 50% centrais dos dados e, por não depender dos extremos, é pouco afetado por valores discrepantes.

```python
A = np.percentile(ARRAY, 75) - np.percentile(ARRAY, 25)
A
```

### Valores discrepantes (ouliers):

Para identificação de outliers devemos construir intervalos de valores a partir da amostra, para isso utilizaremos os quartis. Assim, teremos que calcular o intervalo interquatil (A).

Existem duas formas de outliers, os demominados outlier moderado e o outlier extremo, o seu calculo é dado por:

a) Outilier moderado
OIM = Q1 - 1,5 * A
OSM = Q3 + 1,5 * A

b) Outilier extremo
OIE = Q1 - 3 * A
OSE = Q3 + 3 * A

Os valores de outliers delimitam o linha de fim dos gráficos de boxplot, o que define os valores 'válidos' como >= OI e <=OS.

```python
OIM = np.percentile(ARRAY, 25) - 1.5 * (np.percentile(ARRAY, 75) - np.percentile(ARRAY, 25))
OIM # Outlier inferior moderado
```

```python
OSM = np.percentile(ARRAY, 75) + 1.5 * (np.percentile(ARRAY, 75) - np.percentile(ARRAY, 25))
OSM # Outlier superior moderado
```

```python
OIE = np.percentile(ARRAY, 25) - 3 * (np.percentile(ARRAY, 75) - np.percentile(ARRAY, 25))
OIE # Outlier inferior extremo
```

```python
OSE = np.percentile(ARRAY, 75) + 3 * (np.percentile(ARRAY, 75) - np.percentile(ARRAY, 25))
OSE # Outlier superior extremo
```
